## Reflection

The assignment helped me understand how to develop, compare, and evaluate analytical models. I learned how to use adjusted R², p-values, and residual plots to assess regression performance and how comparing full and reduced models can improve interpretability. I also found StackGP interesting because it demonstrated how symbolic regression can identify mathematical relationships directly from data. These exercises highlighted the importance of balancing model accuracy, complexity, and interpretability.

I can apply these concepts to my individual project on analyzing right-turn crash severity using roadway geometric and traffic characteristics. Although my project focuses on crash severity classification rather than regression, comparing different models and evaluating their performance will be equally important. I plan to use measures such as accuracy, precision, recall, F1-score, and confusion matrices to assess model performance and identify misclassified crash severity outcomes. I can also compare models with different combinations of explanatory variables to determine whether additional roadway and traffic characteristics improve prediction. Overall, this assignment helped me better understand the importance of model selection, evaluation, and validation in developing reliable and interpretable crash severity prediction models.
