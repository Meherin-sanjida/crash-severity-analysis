## Reflection

The class exercises helped me understand how to organize notebook code into reusable Python functions and modules, making it easier to maintain and test. I found Git, autoreload, and Ruff particularly useful for tracking changes, updating code efficiently, and maintaining consistent formatting. The Git exercises also improved my understanding of staging, committing, and pushing changes to GitHub.

I plan to apply these practices to my individual project, which analyzes right-turning vehicles' crash severity at intersections due to different geometric conditions of roadway and traffic conditions. Organizing preprocessing, training, evaluation, and visualization into separate modules will make the workflow easier to manage and reuse across datasets. Git and Ruff will also help maintain code quality, track modifications, and improve the overall reproducibility of my research.
